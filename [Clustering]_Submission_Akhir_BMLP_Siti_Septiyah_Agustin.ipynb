# **5. Data Preprocessing**

Pada tahap ini, data preprocessing adalah langkah penting untuk memastikan kualitas data sebelum digunakan dalam model machine learning. Data mentah sering kali mengandung nilai kosong, duplikasi, atau rentang nilai yang tidak konsisten, yang dapat memengaruhi kinerja model. Oleh karena itu, proses ini bertujuan untuk membersihkan dan mempersiapkan data agar analisis berjalan optimal.

Berikut adalah tahapan-tahapan yang bisa dilakukan, tetapi **tidak terbatas** pada:
1. Menghapus atau Menangani Data Kosong (Missing Values)
2. Menghapus Data Duplikat
3. Normalisasi atau Standarisasi Fitur
4. Deteksi dan Penanganan Outlier
5. Encoding Data Kategorikal
6. Binning (Pengelompokan Data)

Cukup sesuaikan dengan karakteristik data yang kamu gunakan yah.

````python
df_selected = df[['Location', 'Total Cases', 'Total Deaths', 'Location Level']]

# 1. Menangani Missing Values
print("\nMenangani missing values...")
df_selected = df_selected.loc[df_selected['Total Cases'].notna() & df_selected['Location'].notna()]

# 2. Menghapus Data Duplikat
print("\nMenghapus data duplikat...")
df_selected = df_selected.drop_duplicates()

# 3. Deteksi dan Penanganan Outlier
print("\nMendeteksi dan menangani outlier...")
df_selected['z_score'] = np.abs(zscore(df_selected['Total Cases']))
df_selected = df_selected[df_selected['z_score'] <= 3]
df_selected.drop(columns=['z_score'], inplace=True)

# 4. Melakukan Encoding pada Data Kategorikal
print("\nMelakukan encoding pada data kategorikal...")
le_location_level = LabelEncoder()
df_selected['Location Level'] = le_location_level.fit_transform(df_selected['Location Level'])
le_location = LabelEncoder()
df_selected['Location'] = le_location.fit_transform(df_selected['Location'])
# 5. Scaling Data Numerik dengan MinMaxScaler
print("\nMelakukan Scaling dengan MinMaxScaler...")
scaler = MinMaxScaler()
df_selected[['Total Cases', 'Total Deaths']] = scaler.fit_transform(df_selected[['Total Cases', 'Total Deaths']])

# 6. Menampilkan Distribusi Data Setelah Scaling
plt.figure(figsize=(8, 5))
sns.histplot(df_selected['Total Cases'], bins=50, kde=True, color='blue', label="Total Cases")
plt.title("Distribusi 'Total Cases' Setelah Scaling")
plt.xlabel("Total Cases (Scaled)")
plt.ylabel("Frekuensi")
plt.legend()
plt.show()

# Menampilkan dataset setelah preprocessing
print("\nDataset setelah preprocessing:")
print(df_selected.head())
````

# **6. Pembangunan Model Clustering**

## **a. Pembangunan Model Clustering**

Pada tahap ini, Anda membangun model clustering dengan memilih algoritma yang sesuai untuk mengelompokkan data berdasarkan kesamaan. Berikut adalah **rekomendasi** tahapannya.
1. Pilih algoritma clustering yang sesuai.
2. Latih model dengan data menggunakan algoritma tersebut.

````python

optimal_k = 4 
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)

X = df_selected[['Total Cases', 'Total Deaths', 'Location Level', 'Location']]
df_selected['Cluster'] = kmeans.fit_predict(X)
````

## **b. Evaluasi Model Clustering**

Untuk menentukan jumlah cluster yang optimal dalam model clustering, Anda dapat menggunakan metode Elbow atau Silhouette Score.

Metode ini membantu kita menemukan jumlah cluster yang memberikan pemisahan terbaik antar kelompok data, sehingga model yang dibangun dapat lebih efektif. Berikut adalah **rekomendasi** tahapannya.
1. Gunakan Silhouette Score dan Elbow Method untuk menentukan jumlah cluster optimal.
2. Hitung Silhouette Score sebagai ukuran kualitas cluster.

````python
wcss = []  
silhouette_scores = []  
K_range = range(2, 11)  

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X)
        
    wcss.append(kmeans.inertia_)
     
    silhouette_avg = silhouette_score(X, labels)
    silhouette_scores.append(silhouette_avg)

plt.figure(figsize=(8, 5))
plt.plot(K_range, wcss, marker='o', linestyle='--', color='b')
plt.xlabel('Jumlah Cluster (K)')
plt.ylabel('WCSS')
plt.title('Elbow Method untuk Menentukan Jumlah Cluster')
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(K_range, silhouette_scores, marker='o', linestyle='--', color='r')
plt.xlabel('Jumlah Cluster (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score untuk Menentukan Jumlah Cluster')
plt.show()

silhouette_avg = silhouette_score(X, df_selected['Cluster'])
print(f"Silhouette Score untuk K={optimal_k}: {silhouette_avg:.4f}")
````

## **c. Feature Selection (Opsional)**

Silakan lakukan feature selection jika Anda membutuhkan optimasi model clustering. Jika Anda menerapkan proses ini, silakan lakukan pemodelan dan evaluasi kembali menggunakan kolom-kolom hasil feature selection. Terakhir, bandingkan hasil performa model sebelum dan sesudah menerapkan feature selection.

````python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=4)  
X_pca = pca.fit_transform(X_scaled)

df_pca = pd.DataFrame(X_pca, columns=['Total Cases', 'Total Deaths', 'Location Level', 'Location'])

optimal_k = 3  
kmeans_pca = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
df_pca['Cluster'] = kmeans_pca.fit_predict(X_pca)

silhouette_before = silhouette_score(X_scaled, df_selected['Cluster'])

silhouette_after = silhouette_score(X_pca, df_pca['Cluster'])

print(f"Silhouette Score Sebelum Feature Selection: {silhouette_before:.4f}")
print(f"Silhouette Score Sesudah Feature Selection (PCA): {silhouette_after:.4f}")

if silhouette_after > silhouette_before:
    print("✅ Model meningkat setelah feature selection!")
else:
    print("❌ Model lebih baik tanpa feature selection.")
````

Setelah model clustering dilatih dan jumlah cluster optimal ditentukan, langkah selanjutnya adalah menampilkan hasil clustering melalui visualisasi.

````python
X_numeric = df[['Total Cases', 'Total Deaths']] 
scaler.fit(X_numeric) 
df_pca[['Total Cases', 'Total Deaths']] = scaler.inverse_transform(
    df_pca[['Total Cases', 'Total Deaths']]
)
````

Setelah melakukan clustering, langkah selanjutnya adalah menganalisis karakteristik dari masing-masing cluster berdasarkan fitur yang tersedia.

Berikut adalah **rekomendasi** tahapannya.
1. Analisis karakteristik tiap cluster berdasarkan fitur yang tersedia (misalnya, distribusi nilai dalam cluster).
2. Berikan interpretasi: Apakah hasil clustering sesuai dengan ekspektasi dan logika bisnis? Apakah ada pola tertentu yang bisa dimanfaatkan?

````python
df_pca["Cluster"] = df_pca["Cluster"].astype(int)

numeric_cols = df_pca.select_dtypes(include=['number']).columns
cluster_analysis = df_pca.groupby('Cluster')[numeric_cols].mean()

print("\nRata-rata fitur dalam tiap cluster:")
print(cluster_analysis)

cluster_summary = df_pca.groupby('Cluster').agg({
    'Total Cases': ['mean', 'min', 'max'],
    'Total Deaths': ['mean', 'min', 'max']
})

def interpret_cluster(row):
    cases_mean = row[('Total Cases', 'mean')]
    deaths_mean = row[('Total Deaths', 'mean')]
    
    if cases_mean > df_pca['Total Cases'].mean() and deaths_mean > df_pca['Total Deaths'].mean():
        analysis = "jumlah kasus dan kematian tinggi"
    elif cases_mean > df_pca['Total Cases'].mean():
        analysis = "jumlah kasus tinggi, kematian rendah"
    elif deaths_mean > df_pca['Total Deaths'].mean():
        analysis = "jumlah kasus rendah, kematian tinggi"
    else:
        analysis = "jumlah kasus dan kematian rendah"

    return f"Rata-rata Total Cases: {cases_mean:,.2f}\n" \
           f"Rata-rata Total Deaths: {deaths_mean:,.2f}\n" \
           f"Analisis: Cluster ini mencakup wilayah dengan *{analysis}*"

print("\nInterpretasi Hasil Clustering:")
for cluster, row in cluster_summary.iterrows():
    print(f"\n🔹 Cluster {cluster}:")
    print(interpret_cluster(row))
````
